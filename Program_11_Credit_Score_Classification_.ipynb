```python

# Credit Score Classification

import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder

# Sample dataset
data = {
    'Income': [30000, 50000, 70000, 25000, 80000, 45000, 90000, 35000],
    'Loan': [10000, 20000, 15000, 25000, 10000, 30000, 20000, 15000],
    'Late_Payments': [5, 2, 0, 6, 0, 3, 1, 4],
    'Credit_Score': ['Poor', 'Standard', 'Good', 'Poor',
                     'Good', 'Standard', 'Good', 'Poor']
}

df = pd.DataFrame(data)

# Separate input and output
X = df[['Income', 'Loan', 'Late_Payments']]
y = df['Credit_Score']

# Train the model
model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

# Predict credit score
income = 60000
loan = 15000
late_payments = 1

prediction = model.predict([[income, loan, late_payments]])

print("Credit Score Classification")
print("Income:", income)
print("Loan Amount:", loan)
print("Late Payments:", late_payments)
print("Predicted Credit Score:", prediction[0])
```
